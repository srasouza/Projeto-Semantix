#### 8.3.7 — Verificação da consistência entre variáveis

Após verificar individualmente as variáveis quantitativas, será analisada a consistência entre os valores registrados em diferentes colunas que representam informações relacionadas ao mesmo acidente.

Nesta etapa, será verificado se a quantidade total de `feridos` corresponde à soma de `feridos_leves` e `feridos_graves`. Também será analisado se a quantidade de `pessoas` corresponde à soma de `mortos`, `feridos_leves`, `feridos_graves`, `ilesos` e `ignorados`.

Essa verificação permite identificar possíveis inconsistências lógicas nos registros, mesmo quando os valores de cada variável, analisados isoladamente, estão preenchidos corretamente.

A análise será realizada separadamente para 2023, 2024 e 2025, permitindo avaliar se essas relações permanecem consistentes nos três conjuntos de dados antes da consolidação das bases.

```python
# 8.3.7 - Verificação de consisitências entre variáveis

for ano, df in [('2023', df_23),('2024', df_24),('2025', df_25)]:
  # Verificação: feridos = feridos leves + feridos graves
  inconsistencias_feridos = (
      df['feridos'] != (df['feridos_leves'] + df['feridos_graves'])
  ).sum()
  # Verificação: pessoas = mortos + feridos leves + feridos graves + ilesos + ignorados
  inconsistencias_pessoas = (
      df['pessoas'] != (
          df['mortos']
          + df['feridos_leves']
          + df['feridos_graves']
          + df['ilesos']
          + df['ignorados']
      )
  ).sum()

  print('\n--- Ano ---')
  print('Inconsistências em feridos:', inconsistencias_feridos)
  print('Inconsistências em pessoas:', inconsistencias_pessoas)
```

#### 8.3.7.1 — Resultado inicial da verificação de consistência

A verificação das relações entre as variáveis quantitativas mostrou que a variável `feridos` apresenta consistência nos três anos analisados. Em todos os registros, o total de feridos corresponde à soma de `feridos_leves` e `feridos_graves`.

Entretanto, foram identificadas diferenças entre a variável `pessoas` e a soma das variáveis `mortos`, `feridos_leves`, `feridos_graves`, `ilesos` e `ignorados`. Foram encontrados 3.767 registros com diferença em 2023, 3.884 em 2024 e 3.823 em 2025.

Como essas diferenças não indicam, por si só, que os registros estejam incorretos, será realizada uma investigação adicional para identificar a magnitude e o padrão dessas diferenças. Essa análise permitirá compreender se as ocorrências representam inconsistências nos dados ou alguma característica da forma como as informações foram registradas na base.

Nenhum dado será alterado nesta etapa. O tratamento, caso seja necessário, será definido somente após a compreensão da origem dessas diferenças.


```python
# Investigação das inconsistências na variável pessoas

for ano, df in [('2023', df_23), ('2024', df_24), ('2025', df_25)]:

    diferenca = (
        df['pessoas']
        - (
            df['mortos']
            + df['feridos_leves']
            + df['feridos_graves']
            + df['ilesos']
            + df['ignorados']
        )
    )

    print(f'\n--- {ano} ---')
    print('Quantidade de inconsistências:', (diferenca != 0).sum())
    print('Diferenças encontradas:')
    print(diferenca[diferenca != 0].value_counts().sort_index())
```

```python
# Investigação dos registros com inconsistência em pessoas

for ano, df in [('2023', df_23), ('2024', df_24), ('2025', df_25)]:

    soma_pessoas = (
        df['mortos']
        + df['feridos_leves']
        + df['feridos_graves']
        + df['ilesos']
        + df['ignorados']
    )

    diferenca = df['pessoas'] - soma_pessoas

    print(f'\n--- {ano} ---')

    print('\nExemplos com diferença de -1:')
    print(
        df.loc[
            diferenca == -1,
            ['id', 'pessoas', 'mortos', 'feridos_leves',
             'feridos_graves', 'ilesos', 'ignorados']
        ].head(10)
    )

    print('\nExemplos com maiores diferenças:')
    print(
        df.loc[
            diferenca <= -10,
            ['id', 'pessoas', 'mortos', 'feridos_leves',
             'feridos_graves', 'ilesos', 'ignorados']
        ].head(10)
    )
```

#### 8.3.7.2 — Resultado da análise de consistência entre variáveis

Inicialmente, também foi verificada a relação entre `pessoas` e a soma de `mortos`, `feridos_leves`, `feridos_graves`, `ilesos` e `ignorados`. Foram identificadas diferenças em parte dos registros. A investigação dos casos mostrou que essas diferenças estão relacionadas principalmente à variável `ignorados`, que apresenta valores que não devem ser interpretados como uma quantidade de pessoas a ser somada diretamente às demais categorias para reconstrução do total de `pessoas`.

Dessa forma, não serão realizadas alterações nessas variáveis. Os valores originais da fonte serão preservados, evitando a aplicação de uma regra de tratamento sem justificativa sobre o significado da variável `ignorados`.

```python
# 8.5.1 — Visão geral dos acidentes

print('Total de acidentes:', len(df))

print('\nAcidentes por ano:')
print(df['data_inversa'].dt.year.value_counts().sort_index())

print('\nTotal de mortos:', df['mortos'].sum())

print('Total de feridos:', df['feridos'].sum())

print('Total de pessoas envolvidas:', df['pessoas'].sum())

print('\nMédia de pessoas por acidente:', round(df['pessoas'].mean(), 2))

print('\nMédia de mortos por acidente:', round(df['mortos'].mean(), 2))

print('\nMédia de feridos por acidente:', round(df['feridos'].mean(), 2))
```

#### 8.5.1.1 — Resultado da visão geral dos acidentes

```python
# 8.5.2 — Acidentes fatais

acidentes_fatais = (df['mortos'] > 0).sum()
acidentes_nao_fatais = (df['mortos'] == 0).sum()

percentual_fatais = acidentes_fatais / len(df) * 100

print('Acidentes fatais:', acidentes_fatais)
print('Acidentes não fatais:', acidentes_nao_fatais)
print('Percentual de acidentes fatais:', round(percentual_fatais, 2), '%')
```

#### 8.5.2 — Identificação dos acidentes fatais

Para complementar a visão geral dos acidentes, será analisada a quantidade de ocorrências que resultaram em pelo menos uma morte.

Um acidente será considerado fatal quando a variável `mortos` apresentar valor maior que zero. Essa classificação permite diferenciar a quantidade de acidentes fatais dos acidentes que não registraram mortes e calcular a proporção de ocorrências fatais no período analisado.

Essa métrica é relevante para avaliar a gravidade das ocorrências e, posteriormente, investigar quais características dos acidentes estão mais associadas à presença de vítimas fatais.

No período de 2023 a 2025, foram identificados **15.290 acidentes fatais**, correspondendo a **7,16% dos 213.451 acidentes analisados**. Os demais **198.161 acidentes**, equivalentes a **92,84%**, não registraram mortes.

A proporção de acidentes fatais será utilizada nas análises posteriores como uma das principais medidas para avaliar a gravidade das ocorrências.



```python
# 8.5.3 — Evolução anual dos acidentes

df['ano'] = df['data_inversa'].dt.year

evolucao_anual = df.groupby('ano').agg(
    acidentes=('id', 'count'),
    mortos=('mortos', 'sum'),
    feridos=('feridos', 'sum')
).reset_index()

acidentes_fatais_ano = (
    df[df['mortos'] > 0]
    .groupby('ano')
    .size()
    .reset_index(name='acidentes_fatais')
)

evolucao_anual = evolucao_anual.merge(
    acidentes_fatais_ano,
    on='ano',
    how='left'
)

evolucao_anual['percentual_fatais'] = (
    evolucao_anual['acidentes_fatais']
    / evolucao_anual['acidentes']
    * 100
)

print(evolucao_anual)
```

#### 8.5.3.1 — Resultado da evolução anual

A análise da distribuição anual permitiu comparar a quantidade de acidentes, mortes, feridos e acidentes fatais entre 2023 e 2025.

Em 2024 foi registrado o maior número de acidentes, com 73.156 ocorrências, além do maior número de mortes (6.160) e feridos (84.526). Em 2025, foram registrados 72.529 acidentes, 6.043 mortes e 83.550 feridos, enquanto 2023 apresentou os menores valores absolutos entre os três anos.

Apesar das variações nos números absolutos, a proporção de acidentes fatais apresentou comportamento bastante estável: 7,17% em 2023, 7,14% em 2024 e 7,18% em 2025.

Dessa forma, os dados indicam que, no período analisado, não houve uma mudança relevante na proporção de acidentes que resultaram em mortes. As diferenças observadas entre os anos estão principalmente relacionadas ao volume total de ocorrências, enquanto a proporção de acidentes fatais permaneceu próxima de 7,1%.

Esse resultado servirá como referência para as próximas análises, que buscarão identificar se determinadas características dos acidentes apresentam maior associação com ocorrências fatais.

```python
# 8.5.4 — Distribuição dos acidentes por estado

acidentes_uf = (
    df.groupby('uf').size().sort_values(ascending=False)
)

print('Quantidade de acidentes por UF')
print(acidentes_uf)
```

#### 8.5.4.1 — Resultado da distribuição dos acidentes por estado

A análise da distribuição dos acidentes por unidade federativa mostrou diferenças relevantes no volume de ocorrências registradas entre os estados.

Minas Gerais apresentou o maior número de acidentes no período, com **27.873 registros**, seguido por Santa Catarina, com **24.366**, e Paraná, com **22.316**. Na sequência aparecem Rio de Janeiro, com **18.392**, Rio Grande do Sul, com **15.041**, e São Paulo, com **14.320** ocorrências.

Os resultados indicam uma concentração maior de registros em determinados estados. Entretanto, o número absoluto de acidentes não deve ser utilizado isoladamente como indicador de maior gravidade, pois diferenças no volume de ocorrências não significam necessariamente maior proporção de acidentes fatais.

Dessa forma, a distribuição por estado será utilizada como ponto de partida para análises posteriores que relacionem o volume de acidentes com indicadores de gravidade, como mortes e acidentes fatais.

```python
# 8.5.5 — Acidentes e gravidade por UF

acidentes_uf = (
    df.groupby('uf')
    .agg(
        acidentes=('id', 'count'),
        mortos=('mortos', 'sum')
    )
    .reset_index()
)

acidentes_fatais_uf = (
    df[df['mortos'] > 0]
    .groupby('uf')
    .size()
    .reset_index(name='acidentes_fatais')
)

acidentes_uf = acidentes_uf.merge(
    acidentes_fatais_uf,
    on='uf',
    how='left'
)

acidentes_uf['percentual_fatais'] = (
    acidentes_uf['acidentes_fatais']
    / acidentes_uf['acidentes']
    * 100
)

acidentes_uf = acidentes_uf.sort_values(
    'percentual_fatais',
    ascending=False
)

print(acidentes_uf)
```

#### 8.5.5.1 — Resultado da análise de acidentes e gravidade por UF

A análise conjunta do volume de acidentes e dos indicadores de gravidade mostrou que a concentração de ocorrências não acompanha necessariamente a proporção de acidentes fatais.

O Maranhão apresentou a maior proporção de acidentes fatais entre os estados analisados, com **19,59%**, seguido pelo Pará, com **18,92%**. A Bahia também apresentou uma proporção elevada, com **11,61%** de acidentes fatais.

Por outro lado, estados com grande volume de ocorrências apresentaram proporções menores. Minas Gerais, que registrou o maior número de acidentes no período, apresentou **6,80%** de acidentes fatais. São Paulo apresentou **4,42%**, enquanto Santa Catarina apresentou **4,34%**.

Esse resultado evidencia a importância de analisar conjuntamente o volume de acidentes e sua gravidade. Um estado pode apresentar grande quantidade de ocorrências sem necessariamente apresentar a maior proporção de acidentes fatais.

Os estados com menor quantidade de registros devem ser interpretados com cautela, pois volumes reduzidos podem gerar maiores variações nas proporções calculadas. Essa consideração será levada em conta nas análises e na interpretação dos resultados.

Inicialmente, será considerado o número absoluto de ocorrências por causa. Em uma etapa posterior, as principais causas poderão ser relacionadas à ocorrência de acidentes fatais, permitindo avaliar se determinadas causas apresentam maior participação entre as ocorrências com mortes.

Essa análise contribui para a identificação de padrões que poderão ser utilizados na elaboração dos insights e das visualizações do projeto.

```python
# 8.5.6 — Principais causas dos acidentes

causas_acidentes = (
    df['causa_acidente']
    .value_counts()
)

print('Quantidade de de acidentes por causa:')
print(causas_acidentes)
```

#### 8.5.6.1 — Resultado da análise das principais causas

A análise da variável `causa_acidente` identificou **75 categorias diferentes** de causas registradas nos acidentes analisados.

As causas com maior número de ocorrências foram **"Reação tardia ou ineficiente do condutor"**, com 31.574 registros, e **"Ausência de reação do condutor"**, com 31.459 registros. Em seguida, destaca-se **"Acessar a via sem observar a presença dos outros veículos"**, com 20.380 ocorrências.

Também apresentaram volumes relevantes as causas **"Condutor deixou de manter distância do veículo da frente"**, com 13.145 registros, e **"Velocidade Incompatível"**, com 12.675.

Os resultados mostram que algumas categorias concentram uma parcela significativa dos acidentes. Entretanto, a frequência de uma causa não indica necessariamente que ela esteja associada a maior gravidade. Por esse motivo, as principais causas serão posteriormente analisadas em relação à ocorrência de acidentes fatais.

```python
# 8.5.7 — Principais causas e acidentes fatais

top_causas_10 = df['causa_acidente'].value_counts().head(10).index

causas_gravidade = (
    df[df['causa_acidente'].isin(top_causas_10)]
    .groupby('causa_acidente')
    .agg(
        acidentes=('id', 'count'),
        mortos=('mortos', 'sum'),
        acidentes_fatais=('mortos', lambda x: (x > 0).sum())
    ).reset_index()
)

causas_gravidade['percentual_fatais'] = (
    causas_gravidade['acidentes_fatais'] / causas_gravidade['acidentes'] * 100)

causas_gravidade = causas_gravidade.sort_values(
    'percentual_fatais',
    ascending=False
)

print(causas_gravidade)
```

#### 8.5.7.1 — Resultado da análise das principais causas e gravidade

A análise das dez causas de acidentes mais frequentes permitiu comparar o volume de ocorrências com indicadores de gravidade.

A causa **"Reação tardia ou ineficiente do condutor"** apresentou o maior número de acidentes, com 31.574 registros, seguida por **"Ausência de reação do condutor"**, com 31.459. Apesar da elevada frequência, essas causas apresentaram proporções de acidentes fatais de 5,09% e 6,57%, respectivamente.

Em contraste, **"Transitar na contramão"** apresentou 7.300 acidentes, mas registrou a maior proporção de acidentes fatais entre as dez principais causas, com **28,96%**. Também se destacaram **"Velocidade Incompatível"**, com 8,95%, e **"Condutor Dormindo"**, com 7,22%.

Os resultados demonstram que as causas mais frequentes não são necessariamente aquelas associadas às maiores proporções de acidentes fatais. Dessa forma, a análise conjunta de frequência e gravidade permite uma compreensão mais adequada dos padrões encontrados na base.
