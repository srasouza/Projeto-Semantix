### 8.1.2 Distribuição dos registros com valores ausentes

Será verificado se os valores ausentes estão concentrados nos mesmos registros ou distribuídos entre diferentes ocorrências.

```python
print('Registros com pelo menos um valor ausente:')
print('2023:', df_23.isnull().any(axis=1).sum())
print('2024:',df_24.isnull().any(axis=1).sum())
print('2025:', df_25.isnull().any(axis=1).sum())

print('Percentual de registros com pelo menos um valor ausente:')
print('2023:', round(df_23.isnull().any(axis=1).mean() * 100,3),'%')
print('2024:', round(df_24.isnull().any(axis=1).mean() * 100,3), '%')
print('2025:', round(df_25.isnull().any(axis=1).mean() * 100,3), '%')
```

### 8.3.1 Resultado da Verificação das variáveis numéricas

Foram analisadas as variáveis id, latitude, longitude e km, considerando seus tipos de dados, valores nulos e formatos de representação.

A variável id apresentou tipo float64 apenas em 2024, enquanto nos demais anos foi carregada como int64. A verificação mostrou que nenhum identificador possui parte decimal, indicando uma diferença de representação dos dados, e não uma característica dos valores.

As variáveis latitude e longitude não apresentaram valores nulos. Entretanto, nos arquivos de 2023 e 2025 os valores utilizam vírgula como separador decimal e foram carregados como texto, enquanto em 2024 já estão em formato numérico.

A variável km também não apresentou valores nulos. Apesar de estar armazenada como texto nos três anos, os valores encontrados possuem representação numérica, utilizando vírgula como separador decimal em parte dos registros.

Dessa forma, será necessário padronizar os formatos dessas variáveis antes da consolidação dos dados, garantindo que os três anos utilizem os mesmos tipos de dados.

### 8.3.1 — Padronização do id

```python
from IPython.core.shellapp import pt_inputhooks
df_23['id'] = df_23['id'].astype('int64')
df_24['id'] = df_24['id'].astype('int64')
df_25['id'] = df_25['id'].astype('int64')

print('Tipo de Dados')
print('2023:', df_23['id'].dtype)
print('2024:', df_24['id'].dtype)
print('2025:', df_25['id'].dtype)
```

### 8.3.3 — Padronização de latitude e longitude

```python
# Padronização da latitude

df_23['latitude'] = pd.to_numeric(
    df_23['latitude'].str.replace(',', '.', regex=False)
)

df_24['latitude'] = pd.to_numeric(df_24['latitude'])

df_25['latitude'] = pd.to_numeric(
    df_25['latitude'].str.replace(',', '.', regex=False)
)

# Padronização da longitude

df_23['longitude'] = pd.to_numeric(
    df_23['longitude'].str.replace(',', '.', regex=False)
)

df_24['longitude'] = pd.to_numeric(df_24['longitude'])

df_25['longitude'] = pd.to_numeric(
    df_25['longitude'].str.replace(',', '.', regex=False)
)

print('\nTipos após tratamento:')
print('--- Latitude ---')
print('2023:', df_23['latitude'].dtype)
print('2024:', df_24['latitude'].dtype)
print('2025:', df_25['latitude'].dtype)
print('\n--- Longitude ---')
print('2023:', df_23['longitude'].dtype)
print('2024:', df_24['longitude'].dtype)
print('2025:', df_25['longitude'].dtype)
```
